### Descrição dos Alimentos com Mais Calorias por 100g

Aqui estão os alimentos da tabela com o maior teor calórico por 100 gramas, juntamente com suas descrições.

```python
variable_name = "" # @param {"type":"string"}
import pandas as pd

data = {
    'Alimento': [
        'Brócolis',
        'Salmão',
        'Abacate',
        'Quinoa',
        'Mirtilos',
        'Espinafre',
        'Ovos',
        'Amêndoas'
    ],
    'Descrição': [
        'Rico em vitaminas C e K, e fibras.',
        'Fonte de ômega-3 e proteína de alta qualidade.',
        'Rico em gorduras monoinsaturadas saudáveis.',
        'Grão integral rico em proteína e fibras.',
        'Antioxidantes e vitaminas.',
        'Rico em ferro e vitaminas.',
        'Excelente fonte de proteína completa.',
        'Fonte de vitamina E e gorduras saudáveis.'
    ],
    'Calorias_por_100g': [
        55,
        208,
        160,
        120,
        57,
        23,
        155,
        579
    ],
    'Proteina_por_100g': [
        3.7,
        20.4,
        2,
        4.4,
        0.7,
        2.9,
        13,
        21
    ],
    'Carboidratos_por_100g': [
        11.2,
        0,
        8.5,
        21.3,
        14.5,
        3.6,
        1.1,
        22
    ],
    'Gordura_por_100g': [
        0.6,
        13.4,
        14.7,
        1.9,
        0.3,
        0.4,
        11,
        49
    ]
}

df_alimentos_saudaveis = pd.DataFrame(data)

# Encontrar o valor máximo de calorias por 100g
max_calorias = df_alimentos_saudaveis['Calorias_por_100g'].max()

# Filtrar o DataFrame para mostrar os alimentos com o valor máximo de calorias
alimentos_com_mais_calorias = df_alimentos_saudaveis[df_alimentos_saudaveis['Calorias_por_100g'] == max_calorias]

# Exibir apenas as colunas 'Alimento' e 'Descrição'
display(alimentos_com_mais_calorias[['Alimento', 'Descrição', 'Calorias_por_100g']])
```

### Gráfico de Pizza de Carboidratos por 100g de Alimento

Este gráfico de pizza mostra a proporção de carboidratos por 100 gramas em cada alimento, em relação ao total de carboidratos da lista. Ele ajuda a visualizar a contribuição de cada alimento para o teor total de carboidratos.

```python
import matplotlib.pyplot as plt

# Filtrar alimentos com carboidratos > 0 para o gráfico de pizza
df_carboidratos = df_alimentos_saudaveis[df_alimentos_saudaveis['Carboidratos_por_100g'] > 0]

plt.figure(figsize=(10, 8))
plt.pie(
    df_carboidratos['Carboidratos_por_100g'],
    labels=df_carboidratos['Alimento'],
    autopct='%1.1f%%',
    startangle=140,
    pctdistance=0.85
)
plt.title('Proporção de Carboidratos por 100g por Alimento')
plt.axis('equal') # Garante que o gráfico de pizza seja um círculo.
plt.show()
```

### Gráfico de Calorias por 100g de Alimento

Este gráfico de barras mostra a quantidade de calorias por 100 gramas para cada alimento listado na tabela. Isso pode ajudar a ter uma visão rápida da densidade calórica de cada um.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.barplot(x='Alimento', y='Calorias_por_100g', data=df_alimentos_saudaveis, palette='viridis')
plt.title('Calorias por 100g de Alimentos Saudáveis')
plt.xlabel('Alimento')
plt.ylabel('Calorias por 100g')
plt.xticks(rotation=45, ha='right') # Gira os rótulos do eixo X para melhor legibilidade
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout() # Ajusta o layout para evitar sobreposição
plt.show()
```
